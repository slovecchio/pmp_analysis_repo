```python
code = """
-- this is only for NL, there are some doubt in the query expressed 
-- this is only for one day '2026-03-06' 

WITH
  item_price AS (
    -- find the mapping of retail_domain_code with retail_unit
    -- here I hard coded retail_domain_code = 22 because i found it is NL
    -- not sure if it is enough
    SELECT
      calc_version,
      currency_code,
      item_no,
      price as purchase_price,
      valid_from_date,
      valid_to_date
    FROM
      `ingka-dpbsprc-ddc-prod.ingka_retailmixprice_prep.ingka_retailmixprice_prep`
    WHERE
      calc_version = "TERT20262"
      AND retail_domain_code = '22'
      AND currency_code = 'EUR'
      AND (valid_to_date >= '2026-03-06' OR valid_to_date IS NULL)
  ),
  sales_price AS (
    SELECT
      item_no,
      rt.currency_code,
      rt.valid_from_date_time,
      rt.valid_to_date_time,
      rt.price_excl_tax
    FROM
      `ingka-rrm-erix-prod.erix_production.retail_item_sales_price_updated`,
      UNNEST(retail_item_sales_price_list) AS rt
    WHERE
       class_unit_code = 'NL'  -- this correspond to class_unit_type = BU
      -- not sure why the following filter do not apply
      -- class_unit_code = '274'
      -- AND class_unit_type = 'STO'
      AND rt.currency_code = 'EUR'
      AND (
        rt.valid_to_date_time >= '2026-03-06' OR rt.valid_to_date_time IS NULL)
  )

-- this is just for NL store
SELECT
  i.calc_version,
  i.item_no,
  i.currency_code,
  i.purchase_price,
  s.price_excl_tax AS sales_price_excl_tax,
  (s.price_excl_tax - i.purchase_price) AS profit,
  100 * SAFE_DIVIDE(s.price_excl_tax - i.purchase_price, NULLIF(s.price_excl_tax, 0)) AS margin_pct
FROM item_price i
LEFT JOIN sales_price s
USING (item_no)
"""

job=client.query(code)

df_margin=job.to_dataframe()
```

```python
code = """
WITH
  item_location AS (
    SELECT
      article_id,
      bu_code,
      sales_location_id,
      CASE
        WHEN location_type = '1' THEN 'H'
        WHEN location_type = '2' THEN 'P'
        WHEN location_type = '3' THEN 'HP'
        WHEN location_type = '0' THEN 'Addons'
        ELSE 'Other'
        END
        AS location_type_derived,
      /* identify the type of sales location*/
      CASE
        WHEN
          REGEXP_CONTAINS(
            UPPER(TRIM(sales_location_id)),
            r'^M\d{2}[A-Z]\d{2}$')
          THEN 'is_market'
        WHEN
          REGEXP_CONTAINS(
            UPPER(TRIM(sales_location_id)),
            r'^S\d{2}[A-Z]\d{2}$')
          THEN 'is_show'
        WHEN
          REGEXP_CONTAINS(TRIM(sales_location_id), r'^\d{6}$')
          THEN 'is_self_area'
        ELSE 'Other'
        END
        AS type_location_id,
      location_type,
      PARSE_DATE('%Y%m%d', from_date) AS from_date,
      # nex import_date is the until_date with the condition of row_numb = 1
      import_date AS till_date,
      active,
      ROW_NUMBER()
        OVER (
          PARTITION BY article_id, bu_code, sales_location_id, from_date
          ORDER BY CAST(import_date AS DATE) DESC
        ) AS row_num
    FROM
      `ingka-cff-slm-prod.slm.article_location_weekly`
    WHERE
      bu_code = "274"
      AND import_date >= '2026-01-01'
      AND import_date < '2026-03-07'
      AND active = '1'
    QUALIFY row_num = 1
  ),
  per_article AS (
    SELECT
      article_id,
      MAX(CASE WHEN location_type = '3' THEN 1 ELSE 0 END) AS has_hp,
      MAX(CASE WHEN location_type != '3' THEN 1 ELSE 0 END) AS has_other_hp,
      MAX(CASE WHEN location_type = '0' THEN 1 ELSE 0 END) AS has_addons,
      SUM(CASE WHEN location_type = '3' THEN 1 ELSE 0 END) AS sum_hp,
      MAX(CASE WHEN type_location_id = 'is_market' THEN 1 ELSE 0 END)
        AS is_market,
      MAX(CASE WHEN type_location_id = 'is_show' THEN 1 ELSE 0 END)
        AS is_show,
      MAX(CASE WHEN type_location_id = 'is_self_area' THEN 1 ELSE 0 END)
        AS is_self_area
    FROM item_location
    GROUP BY article_id
  ),
  article_candidate AS (
    SELECT article_id
    FROM per_article
    WHERE is_market = 1 AND is_show = 0 AND has_other_hp = 0 AND has_addons = 0
  )
-- select the last (at the day in the where) location of those item
SELECT ac.article_id AS article_candidate, il.*
FROM article_candidate AS ac
LEFT JOIN item_location AS il
  USING (article_id)
WHERE till_date = "2026-03-06"
"""

job=client.query(code)

df_one_market_location=job.to_dataframe()
```

```python
##create 4 range for heat value and merge with df_filtered
import numpy as np
import pandas as pd
import geopandas as gpd
import folium
from shapely.geometry import Point
from shapely import affinity
from folium.plugins import HeatMap

FLOOR_GEOJSON = "./input_geojson_ariadne/274_final_floor1.geojson"
HFB_TARGET = "HFB 11"
EXCLUDE_PA = ["PA 1132","PA 1131"]
INCLUDE_PA = "PA 1111"

ROWS = list("LMNOP")
COLS = list(range(13, 7, -1))   # 16..10
RADIUS_M = 7.0
USE_HEATMAP_LAYER = True

# boundary rows/cols to ALWAYS keep
FORCE_ROWS = {"L", "P"}
FORCE_COLS = {13, 8}

INNER_MARGIN_M  = 3.5
OUTER_MARGIN_M  = 2.0
SHIFT_X_M       = 2.0
SHIFT_Y_M       = 1.5
ROTATE_DEG_CW   = 2.5

# -----------------------
# 1) Read geojson
# -----------------------
gdf = gpd.read_file(FLOOR_GEOJSON)
gdf = gdf.set_crs(4326) if gdf.crs is None else gdf.to_crs(4326)

gdf_poly = gdf[gdf.geometry.type.isin(["Polygon", "MultiPolygon"])].copy()
if "parent_1" not in gdf_poly.columns:
    raise KeyError("Column 'parent_1' not found in geojson.")
if "name" not in gdf_poly.columns:
    raise KeyError("Column 'name' not found in geojson (needed for PA 1433 / PA 1511).")

# -----------------------
# 2) Build allowed area:
#    (HFB14 union + PA1511) - PA1433
# -----------------------
hfb14 = gdf_poly[
    gdf_poly["parent_1"].astype(str).str.upper().str.contains(HFB_TARGET.upper(), na=False)
].copy()
if hfb14.empty:
    raise ValueError(f"No polygons found with parent_1 containing '{HFB_TARGET}'")

hfb14_poly = hfb14.geometry.union_all()

pa1511 = gdf_poly[gdf_poly["name"].astype(str).str.upper().str.strip().eq(INCLUDE_PA.upper())].copy()

exclude_pas = [p.upper().strip() for p in EXCLUDE_PA]
pa1433 = gdf_poly[
    gdf_poly["name"].astype(str).str.upper().str.strip().isin(exclude_pas)
].copy()

#pa1433 = gdf_poly[gdf_poly["name"].astype(str).str.upper().str.strip().eq(EXCLUDE_PA.upper())].copy()

base_poly = hfb14_poly
if not pa1511.empty:
    base_poly = base_poly.union(pa1511.geometry.union_all())
if not pa1433.empty:
    base_poly = base_poly.difference(pa1433.geometry.union_all())

allowed_poly_m = gpd.GeoSeries([base_poly], crs=4326).to_crs(3857).iloc[0]
hfb14_poly_m   = gpd.GeoSeries([hfb14_poly], crs=4326).to_crs(3857).iloc[0]  # for HFB-only visitors

# -----------------------
# 3) Build 7x6 grid points
# -----------------------
poly_for_grid = allowed_poly_m.buffer(-INNER_MARGIN_M)
if poly_for_grid.is_empty or (not poly_for_grid.is_valid):
    poly_for_grid = allowed_poly_m

minx, miny, maxx, maxy = poly_for_grid.bounds
xs = np.linspace(minx, maxx, len(COLS))
ys = np.linspace(maxy, miny, len(ROWS))

xs = xs + SHIFT_X_M
ys = ys + SHIFT_Y_M

cx = (xs.min() + xs.max()) / 2
cy = (ys.min() + ys.max()) / 2
rot_angle = -ROTATE_DEG_CW  # clockwise

grid_records = []
for r_i, row_letter in enumerate(ROWS):
    for c_i, col_num in enumerate(COLS):
        pt = Point(xs[c_i], ys[r_i])
        pt = affinity.rotate(pt, rot_angle, origin=(cx, cy), use_radians=False)

        inside = allowed_poly_m.contains(pt)
        near_border = allowed_poly_m.buffer(OUTER_MARGIN_M).contains(pt)
        keep = inside or near_border or (row_letter in FORCE_ROWS) or (col_num in FORCE_COLS)

        grid_records.append({
            "row": row_letter,
            "col": col_num,
            "grid_id": f"{row_letter}{col_num}",
            "x_m": pt.x,
            "y_m": pt.y,
            "inside_allowed": inside,
            "near_border": near_border,
            "keep": keep
        })

grid_df = pd.DataFrame(grid_records)
grid_df = grid_df[grid_df["keep"]].copy()

grid_gdf_m = gpd.GeoDataFrame(
    grid_df,
    geometry=gpd.points_from_xy(grid_df["x_m"], grid_df["y_m"]),
    crs=3857
)
grid_gdf = grid_gdf_m.to_crs(4326)
grid_df["lat"] = grid_gdf.geometry.y.values
grid_df["lon"] = grid_gdf.geometry.x.values

# -----------------------
# 4) Trajectories: build DF with hash_id + lat/lon
#    (required for UNIQUE visitors) hereeeeeee
# -----------------------
with open("./traject_file/output_trajectories_all_2026-03-06.json", "r", encoding="utf-8") as f:
    trajectory_coords  = json.load(f)   # 'data' is now a dict or list

# -----------------------
# 4) Trajectories: build DF with hash_id + lat/lon
#    (required for UNIQUE visitors)
# -----------------------
df = pd.DataFrame([{
    "timestamp": p["timestamp"],
    "latitude": p["latitude"],
    "longitude": p["longitude"],
    "floor": p["floor"],    
    "hash_id": p["hash_id"],  # user identifier
    "uncertainty": p["uncertainty"]  # uncertainty

} for p in trajectory_coords])

# Sort by user and timestamp to ensure correct ordering
df = df.sort_values(by=["hash_id", "timestamp"]).reset_index(drop=True)
df = df.loc[df.floor ==1]

traj_df = pd.DataFrame(df)

# normalize column names
traj_df = traj_df.rename(columns={"latitude": "lat", "longitude": "lon"})
required_cols = {"hash_id", "lat", "lon"}
missing = required_cols - set(traj_df.columns)
if missing:
    raise KeyError(f"Missing columns for unique visitors: {missing}. "
                   f"Available: {list(traj_df.columns)}")

traj_df["lat"] = pd.to_numeric(traj_df["lat"], errors="coerce")
traj_df["lon"] = pd.to_numeric(traj_df["lon"], errors="coerce")
traj_df = traj_df.dropna(subset=["hash_id", "lat", "lon"]).copy()

traj_gdf_m = gpd.GeoDataFrame(
    traj_df,
    geometry=gpd.points_from_xy(traj_df["lon"], traj_df["lat"]),
    crs=4326
).to_crs(3857)

# Totals (unique visitors)
total_unique_visitors_floor1 = traj_df["hash_id"].nunique()

# HFB14-only visitors (use polygon contains)
in_hfb = traj_gdf_m.geometry.within(hfb14_poly_m)
total_unique_visitors_hfb14 = traj_gdf_m.loc[in_hfb, "hash_id"].nunique()

print("Total unique visitors floor1:", total_unique_visitors_floor1)
print("Total unique visitors HFB14:", total_unique_visitors_hfb14)

# -----------------------
# 5) Heat value per grid = UNIQUE visitors within radius
# -----------------------
traj_xy = np.vstack([traj_gdf_m.geometry.x.values, traj_gdf_m.geometry.y.values]).T
traj_hash = traj_df["hash_id"].astype(str).values

heat_unique = []

for p in grid_gdf_m.geometry:
    dx = traj_xy[:, 0] - p.x
    dy = traj_xy[:, 1] - p.y
    dist2 = dx * dx + dy * dy
    near_mask = dist2 <= (RADIUS_M ** 2)

    if near_mask.any():
        hv = pd.Series(traj_hash[near_mask]).nunique()
    else:
        hv = 0

    heat_unique.append(int(hv))

grid_df["heat_value"] = heat_unique  # unique visitors near grid

# shares
grid_df["total_unique_visitors_floor1"] = total_unique_visitors_floor1
grid_df["total_unique_visitors_hfb14"] = total_unique_visitors_hfb14

grid_df["share_vs_floor1"] = np.where(
    total_unique_visitors_floor1 > 0,
    grid_df["heat_value"] / total_unique_visitors_floor1,
    np.nan
)

grid_df["share_vs_hfb14"] = np.where(
    total_unique_visitors_hfb14 > 0,
    grid_df["heat_value"] / total_unique_visitors_hfb14,
    np.nan
)

# -----------------------
# 6) Simple 3-level bucket on share_vs_hfb14
#    (cold = 0 or very low; warm; hot)
# -----------------------
s = grid_df["share_vs_hfb14"].fillna(0)

# quantiles computed on non-zero shares (to avoid all zeros dominating)
nonzero = s[s > 0]
if len(nonzero) >= 3:
    q1, q2 = nonzero.quantile([0.33, 0.66]).values
else:
    q1, q2 = 0, 0

def bucket_share(v):
    if v <= 0:
        return "cold"
    if v <= q1:
        return "cold"
    if v <= q2:
        return "warm"
    return "hot"

grid_df["grid_heat_cluster"] = s.apply(bucket_share)

# -----------------------
# 7) Plot map
# -----------------------
hfb14_wgs = gpd.GeoSeries([hfb14_poly], crs=4326)
minx2, miny2, maxx2, maxy2 = hfb14_wgs.total_bounds
center = [(miny2 + maxy2) / 2, (minx2 + maxx2) / 2]

m = folium.Map(location=center, zoom_start=21, tiles="cartodbpositron", control_scale=True)

if USE_HEATMAP_LAYER:
    HeatMap(traj_df[["lat", "lon"]].values.tolist(), radius=10, blur=10, max_zoom=25, name="Heat (all points)").add_to(m)

folium.GeoJson(
    hfb14.to_crs(4326).__geo_interface__,
    name="HFB 14",
    style_function=lambda f: {"fillOpacity": 0.05, "weight": 3, "color": "#111"},
).add_to(m)

for _, r in grid_df.iterrows():
    label = r["grid_id"]
    hv = int(r["heat_value"])
    pct_hfb = r["share_vs_hfb14"] * 100 if pd.notna(r["share_vs_hfb14"]) else None
    cl = r["grid_heat_cluster"]

    folium.CircleMarker(
        location=[r["lat"], r["lon"]],
        radius=7,
        color="#111111",
        weight=2,
        fill=True,
        fill_color="#ffffff",
        fill_opacity=0.9,
        tooltip=f"{label} | uniq_users={hv} | share_hfb={pct_hfb:.2f}% | cluster={cl}"
    ).add_to(m)

    folium.Marker(
        location=[r["lat"], r["lon"]],
        icon=folium.DivIcon(
            html=f"""
            <div style="
                font-size:11px; font-weight:700; color:#111;
                background:rgba(255,255,255,0.75);
                border:1px solid rgba(0,0,0,0.25);
                border-radius:6px; padding:1px 5px;
                ">
                {label}
            </div>
            """
        )
    ).add_to(m)

m.fit_bounds([[miny2, minx2], [maxy2, maxx2]])
folium.LayerControl().add_to(m)

display(grid_df.sort_values(["row", "col"]))
m
```
