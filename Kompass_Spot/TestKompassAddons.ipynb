```python
### PRE PERIOD ####


# === Refilled locations: end-to-end script ===
# Requires: geopandas, pandas, numpy, shapely, pyproj, folium, branca

import pandas as pd
import geopandas as gpd
import numpy as np
import re
import folium
from branca.colormap import LinearColormap
from shapely.geometry import Polygon, MultiPolygon
from shapely.ops import unary_union
from pyproj import Geod

# --------------------
# Paths (edit if needed)
# --------------------
GEO_PATH = "inputMalmo/floor1.geojson"
CSV_PATH = "inputMalmo/Refilled_locations_Malmo_pre.csv"   # or "Refilled locations Malmo post.csv"

# --------------------
# Load & split
# --------------------
gdf_all = gpd.read_file(GEO_PATH)
df_csv  = pd.read_csv(CSV_PATH)

gdf_poly  = gdf_all[gdf_all.geometry.type.isin(["Polygon", "MultiPolygon"])].copy()
gdf_point = gdf_all[gdf_all.geometry.type.eq("Point")].copy()

# CRS sanity
if gdf_poly.crs is None:  gdf_poly  = gdf_poly.set_crs(4326, allow_override=True)
if gdf_point.crs is None: gdf_point = gdf_point.set_crs(gdf_poly.crs, allow_override=True)

# --------------------
# Compute geodesic area (m²) per polygon
# --------------------
geod = Geod(ellps="WGS84")

def geodesic_area_geom(geom):
    if geom is None or geom.is_empty:
        return np.nan
    if not geom.is_valid:
        geom = geom.buffer(0)
        if geom.is_empty:
            return np.nan
    if geom.geom_type == "GeometryCollection":
        parts = [g for g in geom.geoms if isinstance(g, (Polygon, MultiPolygon))]
        if not parts:
            return np.nan
        geom = unary_union(parts)
    area_m2, _ = geod.geometry_area_perimeter(geom)
    return abs(area_m2)

gdf_poly["area_m2"] = gdf_poly.geometry.apply(geodesic_area_geom)

# --------------------
# Detect polygon property carrying 'DEPT_###'
# --------------------
poly_dept_col = next(
    (c for c in gdf_poly.columns
     if c != "geometry" and gdf_poly[c].astype(str).str.contains(r"DEPT_\d{3}", regex=True, na=False).any()),
    None
)
if poly_dept_col is None:
    raise ValueError("No polygon property with a 'DEPT_###' id found.")
gdf_poly["dept_id"] = gdf_poly[poly_dept_col].astype(str)

# --------------------
# Build 6-char Refilled location for points
# --------------------
gdf_point["areaLetter"] = gdf_point["areaLetter"].astype("string").str.strip().str.upper()
gdf_point = gdf_point[gdf_point["areaLetter"].isin(["S", "M"])].copy()
gdf_point["coord_raw"]  = gdf_point["coordinate"].astype("string").str.strip()

# Attach DEPT to points via spatial join
poly_keep = gdf_poly[["dept_id", "geometry"]].rename(columns={"dept_id": "parent_dept"})
pts = gpd.sjoin(gdf_point, poly_keep, how="left", predicate="within").drop(columns=["index_right"])

# Map DEPT -> HFB 2-digit code(s)
dept_map = {
    "DEPT_009":"01","DEPT_016":"02","DEPT_012":"03",
    "DEPT_013":"04,05","DEPT_022":"04","DEPT_023":"05",
    "DEPT_005":"06","DEPT_019":"06","DEPT_018":"06",
    "DEPT_011":"07,70","DEPT_024":"07","DEPT_025":"70",
    "DEPT_010":"08","DEPT_014":"09","DEPT_002":"10",
    "DEPT_008":"11","DEPT_007":"12","DEPT_006":"13",
    "DEPT_003":"14,15","DEPT_020":"14","DEPT_021":"15",
    "DEPT_004":"16","DEPT_026":"17","DEPT_001":"18","DEPT_017":"96",
}
def dept_to_codes(dept):
    m = re.search(r"DEPT_\d{3}", str(dept))
    if not m: return []
    mapped = dept_map.get(m.group(0))
    return [x.strip() for x in mapped.split(",")] if mapped else []

# Build the 6-char code
pts["Refilled location"] = pts.apply(
    lambda r: [f"{r['areaLetter']}{d}{r['coord_raw']}" for d in dept_to_codes(r["parent_dept"])],
    axis=1
)
pts = pts.explode("Refilled location", ignore_index=True)
pts["Refilled location"] = pts["Refilled location"].astype(str).str.upper().str.strip()

# --------------------
# CSV prep & join
# --------------------
df_csv.columns = [c.strip() for c in df_csv.columns]
csv_key = "Refilled location" if "Refilled location" in df_csv.columns else "Refilled_location"
if csv_key not in df_csv.columns:
    raise ValueError("CSV must contain 'Refilled location' (or 'Refilled_location').")
value_col = "Refilled pieces" if "Refilled pieces" in df_csv.columns else df_csv.columns[-1]

df_csv["Refilled location"] = df_csv[csv_key].astype(str).str.upper().str.strip()
df_agg = df_csv.groupby("Refilled location", as_index=False)[value_col].sum()

# Matched points
pts_match = pts.merge(df_agg, on="Refilled location", how="inner")
print("Matched points:", len(pts_match))

# --------------------
# Unmatched CSV locations (start with 'M' and NOT containing 'OW')
# --------------------
csv_keys = set(df_agg["Refilled location"].unique())
geo_keys = set(pts["Refilled location"].unique())
unmatched_keys = csv_keys - geo_keys

df_unmatched = (df_csv[df_csv["Refilled location"].isin(unmatched_keys)].copy())
df_unmatched = df_unmatched[
    df_unmatched["Refilled location"].str.startswith("M", na=False)
    & ~df_unmatched["Refilled location"].str.contains("OW", case=False, na=False)
]
df_unmatched_agg = (df_unmatched.groupby("Refilled location", as_index=False)[value_col]
                    .sum()
                    .sort_values("Refilled location"))
print(f"Unmatched CSV locations (M* and not containing 'OW'): {len(df_unmatched_agg)}")
if not df_unmatched_agg.empty:
    print(df_unmatched_agg.to_string(index=False))

# --------------------
# dept_area table with counts/densities and dept_map codes
# --------------------
# base area per dept
dept_area = (gdf_poly.groupby("dept_id", as_index=False)["area_m2"]
             .sum()
             .rename(columns={"area_m2": "dept_area_m2"}))

# counts (matched locations) + total pieces per dept (from matched only)
loc_by_dept = (pts_match.dropna(subset=["parent_dept"])
               .groupby("parent_dept", as_index=False)
               .agg(n_locations=("Refilled location","nunique"),
                    total_refilled=(value_col,"sum")))

dept_area = dept_area.merge(loc_by_dept, left_on="dept_id", right_on="parent_dept", how="left")
dept_area.drop(columns=["parent_dept"], inplace=True)

# fill & densities
dept_area["n_locations"] = dept_area["n_locations"].fillna(0).astype(int)
dept_area["total_refilled"] = dept_area["total_refilled"].fillna(0.0)
dept_area["density_locations_per_m2"] = np.where(
    dept_area["dept_area_m2"] > 0,
    dept_area["n_locations"] / dept_area["dept_area_m2"],
    np.nan
)
dept_area["density_locations_per_1000m2"] = dept_area["density_locations_per_m2"] * 1000

# attach dept_map codes
dept_area["dept_codes"] = dept_area["dept_id"].astype(str).map(dept_map)
dept_area["dept_codes_list"] = dept_area["dept_codes"].apply(
    lambda s: [x.strip() for x in s.split(",")] if isinstance(s, str) else []
)

# preview
want_cols = ["dept_id","dept_codes","dept_area_m2","n_locations","total_refilled",
             "density_locations_per_m2","density_locations_per_1000m2"]
print("\nDepartment area & density summary:")
print(dept_area[want_cols].sort_values("dept_id").to_string(index=False))

# --------------------
# Map (light tiles) + polygon tooltip with DEPT & area
# --------------------
center = gdf_poly.geometry.unary_union.centroid
m = folium.Map(location=[center.y, center.x], zoom_start=20, tiles="CartoDB Positron", control_scale=True)

# Prepare polygons for tooltip
gdf_poly_wgs84 = gdf_poly.to_crs(4326).copy()
gdf_poly_wgs84["dept_id"] = gdf_poly_wgs84["dept_id"].astype(str)
gdf_poly_wgs84["area_label"] = gdf_poly_wgs84["area_m2"].apply(
    lambda x: f"{x:,.0f} m²".replace(",", ".") if pd.notnull(x) else "n/a"
)

folium.GeoJson(
    data=gdf_poly_wgs84.to_json(),
    name="Departments",
    style_function=lambda _:{ "fillOpacity": 0, "color": "#333333", "weight": 1.0 },
    tooltip=folium.features.GeoJsonTooltip(
        fields=["dept_id", "area_label"],
        aliases=["Department:", "Area:"],
        localize=False,
        labels=True,
        sticky=False,
        max_width=300,
    ),
).add_to(m)

# Points (blue → neutral → red), value = refilled pieces
if not pts_match.empty:
    vmin, vmax = float(pts_match[value_col].min()), float(pts_match[value_col].max())
    vmid = float(pts_match[value_col].median())
    cmap = LinearColormap(colors=["#2c7bb6", "#ffffbf", "#d7191c"], index=[vmin, vmid, vmax], vmin=vmin, vmax=vmax)
    cmap.caption = value_col
    cmap.add_to(m)

    pts_wgs84 = pts_match.to_crs(4326)
    for _, r in pts_wgs84.iterrows():
        v = float(r[value_col])
        folium.CircleMarker(
            location=[r.geometry.y, r.geometry.x],
            radius=7,
            color="#222222",
            weight=1,
            fill=True,
            fill_color=cmap(v),
            fill_opacity=0.95,
            tooltip=f"{r['Refilled location']} — {int(v) if float(v).is_integer() else round(v,2)}",
        ).add_to(m)

folium.LayerControl().add_to(m)
m.save("refilled_locations_map_light.html")
m
```

```python
# === Refilled locations interactive map (light background, high contrast) ===
import pandas as pd
import geopandas as gpd
import re
import folium
from branca.colormap import linear, LinearColormap


GEO_PATH = "inputBarendrecth/floor1.geojson"
#CSV_PATH = "inputBarendrecth/Refilled_locations_NL.csv"
CSV_PATH = "inputBarendrecth/Refilled_locations_NL_Jan_feb_26.csv"#"inputBarendrecth/Refilled locations_1_15Dec.csv" 

# Load & split
gdf_all = gpd.read_file(GEO_PATH)
df_csv  = pd.read_csv(CSV_PATH)
gdf_poly  = gdf_all[gdf_all.geometry.type.isin(["Polygon","MultiPolygon"])].copy()
gdf_point = gdf_all[gdf_all.geometry.type.eq("Point")].copy()

# CRS
if gdf_poly.crs is None:  gdf_poly  = gdf_poly.set_crs(4326, allow_override=True)
if gdf_point.crs is None: gdf_point = gdf_point.set_crs(gdf_poly.crs, allow_override=True)

# Points: keep S/M and use RAW coordinate (no padding)

gdf_point["areaLetter"] = gdf_point["areaLetter"].astype("string").str.strip().str.upper()
gdf_point = gdf_point[gdf_point["areaLetter"].isin(["S","M"])].copy()
gdf_point["coord_raw"]  = gdf_point["coordinate"].astype("string").str.strip()

# Attach DEPT from polygons via spatial join
poly_dept_col = next(
    (c for c in gdf_poly.columns if c != "geometry" and
     gdf_poly[c].astype(str).str.contains(r"DEPT_\d{3}", regex=True, na=False).any()),
    None
)
if poly_dept_col is None:
    raise ValueError("No polygon property with a 'DEPT_###' id found.")
poly_keep = gdf_poly[[poly_dept_col, "geometry"]].rename(columns={poly_dept_col: "parent_dept"})
pts = gpd.sjoin(gdf_point, poly_keep, how="left", predicate="within").drop(columns=["index_right"])

# Map DEPT -> 2-digit code(s)
dept_map = {
    "DEPT_009":"01","DEPT_016":"02","DEPT_012":"03",
    "DEPT_013":"04,05","DEPT_022":"04","DEPT_023":"05",
    "DEPT_005":"06","DEPT_019":"06","DEPT_018":"06",
    "DEPT_011":"07,70","DEPT_024":"07","DEPT_025":"70",
    "DEPT_010":"08","DEPT_014":"09","DEPT_002":"10",
    "DEPT_008":"11","DEPT_007":"12","DEPT_006":"13",
    "DEPT_003":"14,15","DEPT_020":"14","DEPT_021":"15",
    "DEPT_004":"16","DEPT_026":"17","DEPT_001":"18","DEPT_017":"96",
}
def dept_to_codes(dept):
    m = re.search(r"DEPT_\d{3}", str(dept))
    if not m: return []
    mapped = dept_map.get(m.group(0))
    return [x.strip() for x in mapped.split(",")] if mapped else []

# Build 6-char code with RAW coordinate
pts["Refilled_location"] = pts.apply(
    lambda r: [f"{r['areaLetter']}{d}{r['coord_raw']}" for d in dept_to_codes(r["parent_dept"])],
    axis=1
)
pts = pts.explode("Refilled_location", ignore_index=True)
pts["Refilled_location"] = pts["Refilled_location"].astype(str).str.upper().str.strip()

# CSV prep & join
df_csv.columns = [c.strip() for c in df_csv.columns]
if "Refilled_location" not in df_csv.columns:
    raise ValueError("CSV must contain 'Refilled_location'.")
value_col = "Refilled_pieces" if "Refilled_pieces" in df_csv.columns else df_csv.columns[-1]
df_csv["Refilled_location"] = df_csv["Refilled_location"].astype(str).str.upper().str.strip()
df_agg = df_csv.groupby("Refilled_location", as_index=False)[value_col].sum()
pts_match = pts.merge(df_agg, on="Refilled_location", how="inner")
print("Matched points:", len(pts_match))
######################### HERE #########################
#########################

# --- Unmatched analysis ---

# Set of all point codes generated from GeoJSON
geo_codes = set(pts["Refilled_location"].dropna().unique())

# 0) Specific request: CSV codes starting with 'M' and NOT 'M00' that are NOT in GeoJSON
mask_M_not00 = df_agg["Refilled_location"].str.match(r"^M(?!00)", na=False)
unmatched_csv_M = df_agg.loc[mask_M_not00 & ~df_agg["Refilled_location"].isin(geo_codes) ,
                             ["Refilled_location", value_col]].drop_duplicates()
print(f"[CSV→GeoJSON] M-not-00 unmatched count: {len(unmatched_csv_M)}")
print(unmatched_csv_M.head(20))


# 0 bis) Specific request: CSV codes starting with 'M' and NOT 'M00' that are NOT in GeoJSON
mask_M_not00 = df_agg["Refilled_location"].str.match(r"^M(?!00)", na=False)
unmatched_csv_M = df_agg.loc[(mask_M_not00) & (~df_agg["Refilled_location"].isin(geo_codes)) & (df_agg["Refilled_pieces"]>0),
                             ["Refilled_location", value_col]].drop_duplicates()
print(f"[CSV→GeoJSON] M-not-00 unmatched count: {len(unmatched_csv_M)}")
print(unmatched_csv_M.head(20))


# 1) CSV locations that didn't match any generated point code (all codes)
#unmatched_csv = df_agg.loc[~df_agg["Refilled_location"].isin(pts["Refilled_location"]), :].copy()
#print(f"CSV locations with no matching point (all codes): {len(unmatched_csv)}")
#print(unmatched_csv.head(10))

# 2) Generated points (with geometry) that didn't find a CSV location
#unmatched_pts = pts.loc[~pts["Refilled_location"].isin(df_agg["Refilled_location"]), :].copy()
#print(f"Point codes with no matching CSV location: {len(unmatched_pts)}")
#print(unmatched_pts[["Refilled_location","areaLetter","coord_raw","parent_dept"]].head(10))


#########################
#########################
#########################
#########################

# Map (light tiles)
center = gdf_poly.geometry.unary_union.centroid
m = folium.Map(location=[center.y, center.x], zoom_start=20, tiles="CartoDB Positron", control_scale=True)

# Department outlines (dark gray for contrast on white)
folium.GeoJson(
    gdf_poly.to_crs(4326).__geo_interface__,
    name="Departments",
    style_function=lambda _:{ "fillOpacity": 0, "color": "#333333", "weight": 1.0 }
).add_to(m)

# Points with vivid colormap + dark ring for contrast
if not pts_match.empty:

    vmin, vmax = float(pts_match[value_col].min()), float(pts_match[value_col].max())
    vmid = float(pts_match[value_col].median())  
    cmap = LinearColormap(
        colors=["#2c7bb6", "#ffffbf", "#d7191c"],  # blue → neutral → red
        index=[vmin, vmid, vmax],                  # place the neutral at the median
        vmin=vmin, vmax=vmax
    )
    cmap.caption = value_col
    cmap.add_to(m)

    

    pts_wgs84 = pts_match.to_crs(4326)
    for _, r in pts_wgs84.iterrows():
        v = float(r[value_col])
        folium.CircleMarker(
            location=[r.geometry.y, r.geometry.x],
            radius=7,
            color="#222222",           # dark outline
            weight=1,
            fill=True,
            fill_color=cmap(v),
            fill_opacity=0.95,
            tooltip=f"{r['Refilled_location']} — {int(v) if v.is_integer() else round(v,2)}",
        ).add_to(m)

folium.LayerControl().add_to(m)
m.save("refilled_locations_map_light_NL_floor1.html")
m
```

```python
# === Floor map + 2 BIG visible locations (M15V13, M15V15) ===
import pandas as pd
import geopandas as gpd
import re
import folium

GEO_PATH = "inputBarendrecth/floor1.geojson"

gdf_all = gpd.read_file(GEO_PATH)
gdf_poly  = gdf_all[gdf_all.geometry.type.isin(["Polygon","MultiPolygon"])].copy()
gdf_point = gdf_all[gdf_all.geometry.type.eq("Point")].copy()

if gdf_poly.crs is None:  gdf_poly  = gdf_poly.set_crs(4326, allow_override=True)
if gdf_point.crs is None: gdf_point = gdf_point.set_crs(gdf_poly.crs, allow_override=True)

gdf_point["areaLetter"] = gdf_point["areaLetter"].astype("string").str.strip().str.upper()
gdf_point = gdf_point[gdf_point["areaLetter"].isin(["S","M"])].copy()
gdf_point["coord_raw"]  = gdf_point["coordinate"].astype("string").str.strip()

poly_dept_col = next(
    (c for c in gdf_poly.columns if c != "geometry" and
     gdf_poly[c].astype(str).str.contains(r"DEPT_\d{3}", regex=True, na=False).any()),
    None
)
if poly_dept_col is None:
    raise ValueError("No polygon property with a 'DEPT_###' id found.")

poly_keep = gdf_poly[[poly_dept_col, "geometry"]].rename(columns={poly_dept_col: "parent_dept"})
pts = gpd.sjoin(gdf_point, poly_keep, how="left", predicate="within").drop(columns=["index_right"])

dept_map = {
    "DEPT_009":"01","DEPT_016":"02","DEPT_012":"03",
    "DEPT_013":"04,05","DEPT_022":"04","DEPT_023":"05",
    "DEPT_005":"06","DEPT_019":"06","DEPT_018":"06",
    "DEPT_011":"07,70","DEPT_024":"07","DEPT_025":"70",
    "DEPT_010":"08","DEPT_014":"09","DEPT_002":"10",
    "DEPT_008":"11","DEPT_007":"12","DEPT_006":"13",
    "DEPT_003":"14,15","DEPT_020":"14","DEPT_021":"15",
    "DEPT_004":"16","DEPT_026":"17","DEPT_001":"18","DEPT_017":"96",
}
def dept_to_codes(dept):
    m = re.search(r"DEPT_\d{3}", str(dept))
    if not m: return []
    mapped = dept_map.get(m.group(0))
    return [x.strip() for x in mapped.split(",")] if mapped else []

pts["location_code"] = pts.apply(
    lambda r: [f"{r['areaLetter']}{d}{r['coord_raw']}" for d in dept_to_codes(r["parent_dept"])],
    axis=1
)
pts = pts.explode("location_code", ignore_index=True)
pts["location_code"] = pts["location_code"].astype(str).str.upper().str.strip()

target_locs = ["M15V13", "M15V15"]
pts_sel = pts[pts["location_code"].isin(target_locs)].copy()

print("Matched target points:", len(pts_sel))
print(pts_sel[["location_code","areaLetter","coord_raw","parent_dept"]].head())

# --- map ---
gdf_poly_w = gdf_poly.to_crs(4326)
pts_wgs84 = pts_sel.to_crs(4326)

# If we found points, center on them; otherwise center on polygons
if not pts_wgs84.empty:
    center_geom = pts_wgs84.geometry.unary_union.centroid
    m = folium.Map(location=[center_geom.y, center_geom.x], zoom_start=22, tiles="CartoDB Positron", control_scale=True)
else:
    center = gdf_poly_w.geometry.unary_union.centroid
    m = folium.Map(location=[center.y, center.x], zoom_start=20, tiles="CartoDB Positron", control_scale=True)

# Polygons behind
folium.GeoJson(
    gdf_poly_w.__geo_interface__,
    name="Departments",
    style_function=lambda _:{ "fillOpacity": 0, "color": "#333333", "weight": 1.0 }
).add_to(m)

# Big, very visible markers (different colors)
color_map = {"M15V13": "#E6007E", "M15V15": "#00A3E0"}  # magenta / cyan

for _, r in pts_wgs84.iterrows():
    code = r["location_code"]
    c = color_map.get(code, "#FF0000")

    # white halo (draw first)
    folium.CircleMarker(
        location=[r.geometry.y, r.geometry.x],
        radius=14,
        color="#FFFFFF",
        weight=6,
        fill=False,
        opacity=1
    ).add_to(m)

    # colored filled point on top
    folium.CircleMarker(
        location=[r.geometry.y, r.geometry.x],
        radius=10,
        color="#111111",
        weight=2,
        fill=True,
        fill_color=c,
        fill_opacity=1,
        tooltip=code,
        popup=code
    ).add_to(m)

# Fit bounds around the two points so they are guaranteed visible
if len(pts_wgs84) >= 1:
    bounds = pts_wgs84.total_bounds  # [minx, miny, maxx, maxy]
    m.fit_bounds([[bounds[1], bounds[0]], [bounds[3], bounds[2]]])

folium.LayerControl().add_to(m)
#m.save("map_only_M15V13_M15V15.html")
m
```
